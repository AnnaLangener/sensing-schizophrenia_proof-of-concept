### Define Rolling Cross-Validation

```python
def rolling_outer_split(
    n_samples,
    train_size,
    test_size,
    step_size
):
    """
    Yields (train_idx, test_idx)
    """
    splits = []
    start = 0

    while start + train_size + test_size <= n_samples:
        train_idx = range(start, start + train_size)
        test_idx = range(start + train_size,
                         start + train_size + test_size)
        splits.append((list(train_idx), list(test_idx)))
        start += step_size

    return splits

def rolling_inner_split(
    train_indices,
    val_size,
    step_size
):
    """
    Nested split inside outer train window
    """
    splits = []
    start = train_indices[0]
    end = train_indices[-1] + 1

    while start + val_size <= end:
        train_idx = list(range(train_indices[0], start))
        val_idx = list(range(start, start + val_size))

        if len(train_idx) > 0:
            splits.append((train_idx, val_idx))

        start += step_size

    return splits
```

### Model

```python
import torch
import torch.nn as nn

class TinyChronosAdapterHead(nn.Module):
    def __init__(self, embed_dim=768, bottleneck=16, dropout=0.05):
        super().__init__()

        # ---- Adapter ----
        self.adapter_layer_norm = nn.LayerNorm(embed_dim)
        self.adapter_down = nn.Linear(embed_dim, bottleneck)
        self.adapter_relu = nn.ReLU()
        self.adapter_dropout = nn.Dropout(dropout)
        self.adapter_up = nn.Linear(bottleneck, embed_dim)

        # ---- Pool projection ----
        self.pool_proj = nn.Linear(embed_dim * 2, embed_dim)

    def forward(self, x):
        """
        x: (batch, seq_len, embed_dim) OR (seq_len, embed_dim)
        returns: pooled (batch, embed_dim)
        """
        if x.dim() == 2:
            x = x.unsqueeze(0)

        # ---- Adapter block ----
        residual = x
        x = self.adapter_layer_norm(x)
        x = self.adapter_down(x)              # [B, S, bottleneck]
        x = self.adapter_relu(x)
        x = self.adapter_dropout(x)
        x = self.adapter_up(x)                # [B, S, D]
        x = x + residual                      # residual connection

        # ---- Statistical pooling ----
        mean = x.mean(dim=1)
        last = x[:, -1]

        pooled = self.pool_proj(torch.cat([mean, last], dim=-1))

        return pooled


class TinyChronosSensorFusion(nn.Module):
    def __init__(
        self,
        embed_dim=768,
        accel_hparams=None, # kept for downstream compatibility (we removed different heads during training)
        battery_hparams=None,
        gps_hparams=None,
        fusion_hparams=None
    ):
        super().__init__()

        fusion_hparams = fusion_hparams or {}

        # ---- Sensor heads ----

        self.all_head = TinyChronosAdapterHead(
            embed_dim,
            bottleneck=fusion_hparams.get("bottleneck", 16),
            dropout=fusion_hparams.get("dropout", 0.05),
        )

        # ---- Per-head normalization ----

        fusion_dropout = fusion_hparams.get("dropout", 0.05)

        # ---- Fusion prediction head ----

        self.fusion_head = nn.Sequential(
            nn.LayerNorm(embed_dim * 1),
            nn.Linear(embed_dim * 1, 32),
            nn.ReLU(),
            nn.Dropout(fusion_dropout),
            nn.Linear(32, 1)
        )

        self.sigmoid = nn.Sigmoid()


    # ---- Forward pass ----

    def forward(self, combined):

        all_out = self.all_head(combined)

        pred = self.sigmoid(
            self.fusion_head(all_out)
        ).squeeze(-1)

        return pred
```

### Define Loss Function

```python
import torch
import torch.nn as nn


class CCCLoss(nn.Module):
    def __init__(self, eps=1e-8):
        super().__init__()
        self.eps = eps

    def forward(self, pred, target):
        pred = pred.float().view(-1)
        target = target.float().view(-1)

        assert pred.shape == target.shape, (
            f"Shape mismatch: pred {pred.shape}, target {target.shape}"
        )

        pred_mean = torch.mean(pred)
        target_mean = torch.mean(target)

        # Use unbiased=False, especially important for small batches
        pred_var = torch.var(pred, unbiased=False)
        target_var = torch.var(target, unbiased=False)

        cov = torch.mean((pred - pred_mean) * (target - target_mean))

        ccc = (2 * cov) / (
            pred_var
            + target_var
            + (pred_mean - target_mean) ** 2
            + self.eps
        )

        return 1.0 - ccc


class RobustRegressionLoss(nn.Module):
    def __init__(self, alpha=0.5, beta=0.0, delta=1.0):
        super().__init__()

        self.huber = nn.HuberLoss(delta=delta)
        self.ccc = CCCLoss()

        self.alpha = alpha
        self.beta = beta

    def variance_loss(self, x):
        x = x.float().view(-1)
        std = torch.sqrt(torch.var(x, unbiased=False) + 1e-8)
        return torch.relu(0.10 - std)

    def forward(self, pred, target):
        pred = pred.float().view(-1)
        target = target.float().view(-1)

        assert pred.shape == target.shape, (
            f"Shape mismatch: pred {pred.shape}, target {target.shape}"
        )

        huber = self.huber(pred, target)
        ccc = self.ccc(pred, target)

        if self.beta > 0:
            var = self.variance_loss(pred)
            return huber + self.alpha * ccc + self.beta * var

        return huber + self.alpha * ccc
```

### Helper Functions

```python
import copy
import math
import random

import numpy as np
import torch
from sklearn.metrics import mean_squared_error

def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    # Optional stricter reproducibility
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


def pad_ragged_batch_left(x_list, pad_value=np.nan):
    lengths = [x.shape[0] for x in x_list]
    max_len = max(lengths)
    feat_dim = x_list[0].shape[1]

    out = np.full((len(x_list), max_len, feat_dim), pad_value, dtype=np.float32)

    for i, x in enumerate(x_list):
        out[i, -x.shape[0]:, :] = x  # left pad, real data ends at right edge

    return out

def get_labels_for_outcome(labels_all, outcome):
    """
    labels_all shape: [n_samples, 3]
    Columns: negative, cognitive, positive
    """
    if outcome == "negative":
        return labels_all[:, 0]
    elif outcome == "cognitive":
        return labels_all[:, 1]
    elif outcome == "positive":
        return labels_all[:, 2]
    else:
        raise ValueError(f"Unknown outcome: {outcome}")


def evaluate_selected_seed_models(
    seed_runs,
    hparams,
    val_embeds,
    y_val,
    device,
    selection_mode,
    top_k_ensemble,
):
    """
    Evaluate either:
    - single best seed
    - ensemble of top_k seeds

    This is used during inner CV to select the best single model and the best
    ensemble model separately.
    """

    seed_runs = sorted(seed_runs, key=lambda item: item["val_loss"])

    if selection_mode == "single":
        selected_models = seed_runs[:1]
    elif selection_mode == "ensemble":
        selected_models = seed_runs[: min(top_k_ensemble, len(seed_runs))]
    else:
        raise ValueError(f"Unknown selection_mode: {selection_mode}")

    val_predictions = [
        predict_with_state(
            model_info["state"],
            hparams,
            val_embeds,
            device
        )
        for model_info in selected_models
    ]

    selected_val_pred = np.mean(
        np.stack(val_predictions, axis=0),
        axis=0
    )

    selected_val_loss = mean_squared_error(
        np.asarray(y_val).reshape(-1),
        selected_val_pred.reshape(-1)
    )



    return selected_models, selected_val_loss, selected_val_pred
```

### Main Loop

```python
# =============================================================================
# LoRA-only Chronos-2 fine-tuning path
# =============================================================================
from torch.utils.data import Dataset, DataLoader

USE_CHRONOS_LORA = True

if USE_CHRONOS_LORA:
    from peft import (
        LoraConfig,
        get_peft_model,
        get_peft_model_state_dict,
        set_peft_model_state_dict,
    )

    LORA_SETTINGS = {
        "r": 8,
        "alpha": 16,
        "dropout": 0.05,
        "target_modules": ["q", "k", "v", "o"],
        "batch_size": 6,
    }


    AUG_TECHNIQUES = ["none"]

    BASE_CHRONOS_MODEL = copy.deepcopy(pipeline.model).cpu()
    pipeline.model = pipeline.model.cpu()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    def build_lora_config():
        return LoraConfig(
            r=LORA_SETTINGS["r"],
            lora_alpha=LORA_SETTINGS["alpha"],
            lora_dropout=LORA_SETTINGS["dropout"],
            target_modules=LORA_SETTINGS["target_modules"],
            bias="none",
            inference_mode=False,
        )

    def _model_device(module):
        return next(module.parameters()).device

    def _copy_lora_state_dict(state_dict):
        return {
            key: value.detach().cpu().clone()
            for key, value in state_dict.items()
        }

    class CombinedSequenceDataset(Dataset):
        def __init__(self, combined, y):
            self.combined = [np.asarray(item, dtype=np.float32) for item in combined]
            self.y = torch.as_tensor(
                np.asarray(y, dtype=np.float32).reshape(-1),
                dtype=torch.float32,
            )

        def __len__(self):
            return len(self.y)

        def __getitem__(self, idx):
            return {
                "combined": self.combined[idx],
                "y": self.y[idx],
            }

    def collate_combined_sequences(batch):
        return {
            "combined": [item["combined"] for item in batch],
            "y": torch.stack([item["y"] for item in batch]).view(-1),
        }

    def embed_no_output_patch_trainable(
        backbone,
        inputs,
        batch_size: int = 256,
        context_length: int | None = None,
    ):
        core = backbone.get_base_model() if hasattr(backbone, "get_base_model") else backbone
        model_device = _model_device(core)
        model_context_length = core.chronos_config.context_length

        if context_length is None:
            context_length = model_context_length

        if context_length > model_context_length:
            warnings.warn(
                f"context_length {context_length} > model default {model_context_length}; resetting."
            )
            context_length = model_context_length

        test_dataset = Chronos2Dataset(
            inputs,
            context_length=context_length,
            prediction_length=0,
            batch_size=batch_size,
            output_patch_size=core.chronos_config.output_patch_size,
            mode=DatasetMode.TEST,
        )

        test_loader = DataLoader(
            test_dataset,
            batch_size=None,
            num_workers=0,
            pin_memory=model_device.type == "cuda",
            shuffle=False,
            drop_last=False,
        )

        all_embeds = []

        for batch in test_loader:
            batch_context = batch["context"]
            batch_group_ids = batch["group_ids"]
            batch_target_idx_ranges = batch["target_idx_ranges"]

            encoder_outputs, _, _, _ = core.encode(
                context=batch_context.to(device=model_device, dtype=torch.float32),
                group_ids=batch_group_ids.to(model_device),
                num_output_patches=0,
            )

            hidden = encoder_outputs[0]
            batch_embeds = [
                hidden[start:end] for (start, end) in batch_target_idx_ranges
            ]
            all_embeds.extend(batch_embeds)

        return all_embeds

    def embed_stream_trainable(
        backbone,
        x_list,
        drop_special_tokens=False,
        pool_variates=True,
    ):
        x_pad = pad_ragged_batch_left(
            [np.asarray(item, dtype=np.float32) for item in x_list]
        )

        x_t = torch.as_tensor(x_pad, dtype=torch.float32).permute(0, 2, 1)
        emb = embed_no_output_patch_trainable(backbone, x_t)
        out = torch.stack(emb)

        if drop_special_tokens:
            out = out[:, :, :-1, :]

        if pool_variates:
            out = out.mean(dim=1)

        return out

    class TinyChronosLoRARegressor(nn.Module):
        def __init__(self, hparams, device):
            super().__init__()
            backbone = copy.deepcopy(BASE_CHRONOS_MODEL).to(device)
            self.backbone = get_peft_model(backbone, build_lora_config())
            self.backbone.print_trainable_parameters()
            self.head = TinyChronosSensorFusion(
                embed_dim=768,
                fusion_hparams=hparams["fusion"],
            ).to(device)
            self.device = device

        def forward(self, combined):
            if isinstance(combined, torch.Tensor):
                embeds = combined.to(self.device)
            else:
                embeds = embed_stream_trainable(self.backbone, combined)
            return self.head(embeds)

    def extract_trainable_state(model):
        return {
            "head": copy.deepcopy(model.head.state_dict()),
            "lora": _copy_lora_state_dict(
                get_peft_model_state_dict(model.backbone)
            ),
        }

    def load_trainable_state(model, state_bundle):
        model.head.load_state_dict(state_bundle["head"])
        set_peft_model_state_dict(model.backbone, state_bundle["lora"])
        return model


    def load_split_model_state(model, saved_state):
        """
        Load a checkpoint stored as:
            {
                "head": head_state_dict,
                "lora": lora_state_dict,
            }
        """

        if not isinstance(saved_state, dict):
            raise TypeError(
                f"Expected saved_state to be a dict, got {type(saved_state)}"
            )

        # Older/full checkpoints
        if "head" not in saved_state or "lora" not in saved_state:
            model.load_state_dict(saved_state)
            return model

        # Load downstream prediction head
        model.head.load_state_dict(
            saved_state["head"],
            strict=True,
        )

        # Load LoRA adapter weights
        set_peft_model_state_dict(
            model.backbone,
            saved_state["lora"],
            adapter_name="default",
        )

        return model

    def _unpack_batch(batch, device):
        combined = batch["combined"]
        y = batch["y"].to(device).float().view(-1)
        return combined, y

    def train_epoch(model, loader, optimizer, criterion, device, plot=False):
        model.train()
        train_loss = 0.0
        all_preds = []
        all_targets = []

        for batch in loader:
            combined, y = _unpack_batch(batch, device)

            optimizer.zero_grad()
            preds = model(combined).view(-1)
            loss = criterion(preds, y)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            all_preds.append(preds.detach().cpu().numpy())
            all_targets.append(y.detach().cpu().numpy())

        if plot and all_preds:
            all_preds = np.concatenate(all_preds).ravel()
            all_targets = np.concatenate(all_targets).ravel()

            plt.figure(figsize=(10, 3.5))
            plt.plot(all_targets, label="True values", marker='.')
            plt.plot(all_preds, label="Predicted values", marker='.')
            plt.xlabel("Sample index")
            plt.ylabel("Value")
            plt.title("Training: Predicted vs True values (line plot)")
            plt.legend()
            plt.ylim(0, 1)
            plt.style.use('dark_background')
            plt.tight_layout()
            plt.show()

        return train_loss / max(1, len(loader))

    @torch.no_grad()
    def eval_epoch(model, loader, criterion, device, plot=False):
        model.eval()
        total_loss = 0.0
        all_preds = []
        all_targets = []

        for batch in loader:
            combined, y = _unpack_batch(batch, device)
            preds = model(combined).view(-1)
            loss = criterion(preds, y)

            total_loss += loss.item()
            all_preds.append(preds.detach().cpu().numpy())
            all_targets.append(y.detach().cpu().numpy())

        if plot and all_preds:
            all_preds = np.concatenate(all_preds).ravel()
            all_targets = np.concatenate(all_targets).ravel()

            r = np.corrcoef(all_preds, all_targets)[0, 1]
            r2 = r2_score(all_targets, all_preds)

            print("*" * 80)
            print(f"r = {r:.4f}, R² = {r2:.4f}")
            pred_std = all_preds.std()
            true_std = all_targets.std()
            print("variance ratio:", pred_std / true_std)

            plt.figure(figsize=(10, 3.5))
            plt.plot(all_targets, label="True values", marker='o')
            plt.plot(all_preds, label="Predicted values", marker='x')
            plt.xlabel("Sample index")
            plt.ylabel("Value")
            plt.title("Validation: Predicted vs True values (line plot)")
            plt.legend()
            plt.style.use('dark_background')
            plt.ylim(0, 1)
            plt.tight_layout()
            plt.show()

        return total_loss / max(1, len(loader))

    def build_model_from_hparams(hparams, device):
        return TinyChronosLoRARegressor(hparams, device).to(device)

    def predict_with_state(state_dict, hparams, embeds, device):
        model = build_model_from_hparams(hparams, device)
        load_trainable_state(model, state_dict)
        model.eval()
        with torch.no_grad():
            preds = model(embeds["combined"]).cpu().numpy().reshape(-1)
        del model
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
        return preds

    def clone_selected_models(selected_models):
        return copy.deepcopy(selected_models)

    def train_single_seed_model(
        train_loader,
        val_loader,
        hparams,
        device,
        seed,
        max_epochs=100,
        patience=15,
    ):
        seed_everything(seed)
        model = build_model_from_hparams(hparams, device)
        criterion = RobustRegressionLoss(alpha=hparams["alpha"])

        optimizer = optim.AdamW(
              [
                  {
                      "params": model.backbone.parameters(),
                      "lr": hparams["lora_lr"],
                  },
                  {
                      "params": model.head.parameters(),
                      "lr": hparams["head_lr"],
                  },
              ],
              betas=(0.9, 0.95),
          )

        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer,
            mode="min",
            factor=0.5,
            patience=5,
        )

        counter = 0
        best_inner_val = float("inf")
        best_inner_state = None
        best_inner_epoch = None

        for epoch in range(max_epochs):
            plot_now = ((epoch + 1) % 5 == 0)
            epoch_loss = train_epoch(model, train_loader, optimizer, criterion, device, plot=plot_now)

            val_loss = eval_epoch(model, val_loader, criterion, device, plot=plot_now)
            scheduler.step(val_loss)

            print("=" * 80)
            print(
                f"[Seed {seed}] Epoch [{epoch + 1}/{max_epochs}] "
                f"- Train Loss: {epoch_loss:.4f}, Val Loss: {val_loss:.4f}"
            )
            print("=" * 80)

            if math.isfinite(val_loss) and val_loss < best_inner_val:
                best_inner_val = val_loss
                best_inner_state = extract_trainable_state(model)
                best_inner_epoch = epoch + 1
                counter = 0
            else:
                counter += 1
                if counter >= patience:
                    break

        if best_inner_state is None:
            raise RuntimeError(f"No valid model state found for seed {seed}.")

        del model
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        return {
            "seed": seed,
            "val_loss": best_inner_val,
            "state": best_inner_state,
            "best_epoch": best_inner_epoch,
        }

    def retrain_model_on_outer_train(
        train_features_combined,
        y_train,
        best_hparams,
        model_info,
        device,
        retrain_epoch_scale=0.7,
        retrain_lr=1e-4,
        batch_size=None,
        aug_technique="none",
    ):
        model_seed = model_info["seed"]
        retrain_epochs = max(1, int(round(model_info["best_epoch"] * retrain_epoch_scale)))

        final_hparams = copy.deepcopy(best_hparams)
        final_hparams["head_lr"] = retrain_lr

        final_train_ds = CombinedSequenceDataset(train_features_combined, y_train)
        final_monitor_ds = CombinedSequenceDataset(train_features_combined, y_train)

        effective_batch_size = batch_size or LORA_SETTINGS["batch_size"]

        final_train_loader = DataLoader(
            final_train_ds,
            batch_size=effective_batch_size,
            shuffle=True,
            drop_last=False,
            collate_fn=collate_combined_sequences,
        )

        final_monitor_loader = DataLoader(
            final_monitor_ds,
            batch_size=effective_batch_size,
            shuffle=False,
            drop_last=False,
            collate_fn=collate_combined_sequences,
        )

        print(
            f"Retraining seed {model_seed} for {retrain_epochs} epochs "
            f"with retrain_lr={retrain_lr}"
        )

        retrained_result = train_single_seed_model(
            train_loader=final_train_loader,
            val_loader=final_monitor_loader,
            hparams=final_hparams,
            device=device,
            seed=model_seed,
            max_epochs=retrain_epochs,
            patience=retrain_epochs + 1,
        )

        final_model = build_model_from_hparams(final_hparams, device)

        load_split_model_state(
            final_model,
            retrained_result["state"],
        )

        final_model.eval()

        return final_model, retrained_result, final_hparams

    def run_nested_cv_for_person(
        features_combined,
        labels,
        participant_id,
        outcome,
        outer_cfg,
        inner_cfg,
        device,
        hyperparams_grid,
        model_dirs,
        result_dirs,
        embed_dir,
        final_variants,
        max_epochs=100,
        patience=15,
        num_seed_runs=10,
        top_k_ensemble=5,
        retrain_epoch_scale=0.7,
        retrain_lr=1e-4,
        seed_base=42,
        aug_technique="none",
    ):
        selection_modes = sorted(
            set(item["selection_mode"] for item in final_variants)
        )

        outer_splits = rolling_outer_split(
            n_samples=len(labels),
            **outer_cfg,
        )

        print("\n" + "#" * 80)
        print(f"Participant: {participant_id}")
        print(f"Outcome: {outcome}")
        print(f"Selection modes: {selection_modes}")
        print(f"Final variants: {[item['variant_tag'] for item in final_variants]}")
        print(f"Top-k ensemble: {top_k_ensemble}")
        print("#" * 80)

        results_by_variant = {}

        for variant_info in final_variants:
            variant_tag = variant_info["variant_tag"]
            result_dir_now = result_dirs[variant_tag]
            result_path = result_dir_now / f"results_{participant_id}_{outcome}.pth"

            if result_path.exists():
                loaded_results = torch.load(
                    result_path,
                    map_location="cpu",
                    weights_only=False,
                )

                if not isinstance(loaded_results, list):
                    raise TypeError(
                        f"Expected a list in {result_path}, but found {type(loaded_results).__name__}"
                    )

                results_by_variant[variant_tag] = loaded_results
            else:
                results_by_variant[variant_tag] = []

        for outer_i, (train_idx, test_idx) in enumerate(outer_splits):
            completed_for_all_variants = all(
                any(
                    result["outer_i"] == outer_i
                    for result in results_by_variant[variant_info["variant_tag"]]
                )
                for variant_info in final_variants
            )

            if completed_for_all_variants:
                print(f"Skipping completed outer fold {outer_i}")
                continue

            y_train = labels[train_idx]
            y_test = labels[test_idx]

            X_tr_combined = features_combined[train_idx]
            X_test_combined = features_combined[test_idx]

            inner_splits = rolling_inner_split(train_idx, **inner_cfg)
            best_by_selection = {
                selection_mode: {
                    "best_val": float("inf"),
                    "best_hparams": None,
                    "best_seed_models": None,
                    "best_y_val": None,
                    "best_val_features": None,
                    "best_inner_i": None
                }
                for selection_mode in selection_modes
            }

            for inner_i, (inner_train_idx, val_idx) in enumerate(inner_splits):
                y_tr = labels[inner_train_idx]
                y_val = labels[val_idx]
                X_inner_tr_combined = features_combined[inner_train_idx]
                X_inner_val_combined = features_combined[val_idx]

                train_ds = CombinedSequenceDataset(X_inner_tr_combined, y_tr)
                val_ds = CombinedSequenceDataset(X_inner_val_combined, y_val)

                train_loader = DataLoader(
                    train_ds,
                    batch_size=LORA_SETTINGS["batch_size"],
                    shuffle=True,
                    drop_last=False,
                    collate_fn=collate_combined_sequences,
                )

                val_loader = DataLoader(
                    val_ds,
                    batch_size=LORA_SETTINGS["batch_size"],
                    shuffle=False,
                    drop_last=False,
                    collate_fn=collate_combined_sequences,
                )

                val_features = {
                    "combined": list(X_inner_val_combined),
                }

                for h_idx, hparams in enumerate(hyperparams_grid, start=1):
                    print(f"\nHyperparameter set {h_idx}/{len(hyperparams_grid)}")
                    print("  Fusion head:", hparams.get("fusion", {}))
                    print("  LoRA learning rate:", hparams.get("lora_lr"))
                    print("  Head learning rate:", hparams.get("head_lr"))
                    print("  Alpha:", hparams.get("alpha", None))

                    seed_runs = []
                    seed_offset = seed_base + outer_i * 1000 + inner_i * 100 + h_idx * 10
                    seed_candidates = [seed_offset + seed_j for seed_j in range(num_seed_runs)]

                    for seed in seed_candidates:
                        print(f"Running seed {seed} for hyperparameter set {h_idx}")
                        seed_result = train_single_seed_model(
                            train_loader=train_loader,
                            val_loader=val_loader,
                            hparams=hparams,
                            device=device,
                            seed=seed,
                            max_epochs=max_epochs,
                            patience=patience,
                        )
                        seed_runs.append(seed_result)

                    for selection_mode in selection_modes:
                        selected_models, selected_val_loss, selected_val_pred = (
                            evaluate_selected_seed_models(
                                seed_runs=seed_runs,
                                hparams=hparams,
                                val_embeds=val_features,
                                y_val=y_val,
                                device=device,
                                selection_mode=selection_mode,
                                top_k_ensemble=top_k_ensemble,
                            )
                        )

                        if (
                            math.isfinite(selected_val_loss)
                            and selected_val_loss < best_by_selection[selection_mode]["best_val"]
                        ):
                            best_by_selection[selection_mode]["best_val"] = selected_val_loss
                            best_by_selection[selection_mode]["best_hparams"] = copy.deepcopy(hparams)
                            best_by_selection[selection_mode]["best_seed_models"] = clone_selected_models(selected_models)
                            best_by_selection[selection_mode]["best_y_val"] = np.asarray(y_val).copy()
                            best_by_selection[selection_mode]["best_val_features"] = {
                                "combined": list(X_inner_val_combined),
                            }
                            best_by_selection[selection_mode]["best_inner_i"] = inner_i

                if torch.cuda.is_available():
                    torch.cuda.empty_cache()

            for selection_mode in selection_modes:
                if (
                    best_by_selection[selection_mode]["best_hparams"] is None
                    or not best_by_selection[selection_mode]["best_seed_models"]
                ):
                    raise RuntimeError(
                        f"No model selected for participant {participant_id}, outcome {outcome}, outer split {outer_i}, selection mode {selection_mode}."
                    )

            for variant_info in final_variants:
                variant_tag = variant_info["variant_tag"]
                selection_mode = variant_info["selection_mode"]
                retrain_final_now = variant_info["retrain_final"]

                already_completed = any(
                    result["outer_i"] == outer_i
                    for result in results_by_variant[variant_tag]
                )

                if already_completed:
                    print(f"Skipping outer fold {outer_i} for {variant_tag}: result already exists.")
                    continue

                best_info = best_by_selection[selection_mode]
                best_val = best_info["best_val"]
                best_hparams = best_info["best_hparams"]
                best_seed_models = best_info["best_seed_models"]
                best_y_val = best_info["best_y_val"]
                best_val_features = best_info["best_val_features"]
                best_inner_i = best_info["best_inner_i"]

                model_dir_now = model_dirs[variant_tag]
                result_dir_now = result_dirs[variant_tag]

                test_preds = []
                val_preds = []
                final_model_infos = []

                for model_info in best_seed_models:
                    model_seed = model_info["seed"]

                    if retrain_final_now:
                        final_model, retrained_result, final_hparams = retrain_model_on_outer_train(
                            train_features_combined=X_tr_combined,
                            y_train=y_train,
                            best_hparams=best_hparams,
                            model_info=model_info,
                            device=device,
                            retrain_epoch_scale=retrain_epoch_scale,
                            retrain_lr=retrain_lr,
                            batch_size=LORA_SETTINGS["batch_size"],
                            aug_technique=aug_technique,
                        )

                        final_info = {
                            "seed": model_seed,
                            "mode": "retrained",
                            "selection_mode": selection_mode,
                            "original_best_epoch": model_info["best_epoch"],
                            "retrain_epochs": max(1, int(round(model_info["best_epoch"] * retrain_epoch_scale))),
                            "original_inner_val_loss": model_info["val_loss"],
                            "final_hparams": final_hparams,
                        }
                        final_state = retrained_result["state"]
                    else:
                        final_hparams = copy.deepcopy(best_hparams)
                        final_model = build_model_from_hparams(final_hparams, device)
                        load_trainable_state(final_model, model_info["state"])
                        final_model.eval()

                        final_info = {
                            "seed": model_seed,
                            "mode": "selected_inner_model_no_retrain",
                            "selection_mode": selection_mode,
                            "best_epoch": model_info["best_epoch"],
                            "inner_val_loss": model_info["val_loss"],
                            "final_hparams": final_hparams,
                        }
                        final_state = copy.deepcopy(model_info["state"])

                    model_path = (
                        model_dir_now
                        / f"participant_{participant_id}"
                        / outcome
                        / f"outer{outer_i}"
                        / f"seed_{model_seed}.pth"
                    )

                    model_path.parent.mkdir(parents=True, exist_ok=True)
                    torch.save(
                        {
                            "state": final_state,
                            "hparams": final_hparams,
                            "seed": model_seed,
                            "lora_settings": copy.deepcopy(LORA_SETTINGS),
                        },
                        model_path,
                    )

                    with torch.no_grad():
                        member_test_pred = final_model(list(X_test_combined)).cpu().numpy().reshape(-1)
                        member_val_pred = final_model(best_val_features["combined"]).cpu().numpy().reshape(-1)

                    test_preds.append(member_test_pred)
                    val_preds.append(member_val_pred)

                    final_info["model_path"] = str(model_path)
                    final_model_infos.append(final_info)

                    del final_model
                    if torch.cuda.is_available():
                        torch.cuda.empty_cache()

                y_pred = np.mean(np.stack(test_preds, axis=0), axis=0)
                y_pred_val = np.mean(np.stack(val_preds, axis=0), axis=0)

                y_true = np.asarray(y_test).reshape(-1)
                y_pred = np.asarray(y_pred).reshape(-1)
                y_pred_val = np.asarray(y_pred_val).reshape(-1)
                split_mse = mean_squared_error(y_true, y_pred)

                y_train_mean_array = np.full_like(y_true, y_train.mean())
                metadata_path = (
                    model_dir_now
                    / f"participant_{participant_id}"
                    / outcome
                    / f"outer{outer_i}"
                    / "metadata.pth"
                )

                metadata_path.parent.mkdir(parents=True, exist_ok=True)
                torch.save(
                    {
                        "participant_id": participant_id,
                        "outcome": outcome,
                        "outer_split": outer_i,
                        "variant_tag": variant_tag,
                        "selection_mode": selection_mode,
                        "retrain_final": retrain_final_now,
                        "top_k_ensemble": top_k_ensemble,
                        "effective_top_k": 1 if selection_mode == "single" else top_k_ensemble,
                        "best_inner_i": best_inner_i,
                        "best_inner_val_loss": best_val,
                        "best_hparams": best_hparams,
                        "selected_seeds": [model_info["seed"] for model_info in best_seed_models],
                        "final_model_infos": final_model_infos,
                        "lora_settings": copy.deepcopy(LORA_SETTINGS),
                    },
                    metadata_path,
                )

                results_by_variant[variant_tag].append(
                    {
                        "participant_id": participant_id,
                        "outcome": outcome,
                        "variant_tag": variant_tag,
                        "selection_mode": selection_mode,
                        "outer_i": outer_i,
                        "train_idx": np.asarray(train_idx),
                        "test_idx": np.asarray(test_idx),
                        "y_test": y_true,
                        "y_pred": y_pred,
                        "y_val": best_y_val,
                        "y_pred_val": y_pred_val,
                        "y_train": y_train,
                        "y_train_mean": y_train_mean_array,
                        "y_train_last": np.full_like(y_true, y_train[-1]),
                        "split_mse": split_mse,
                        "best_epoch_attn": None,
                        "best_hparams": best_hparams,
                        "best_inner_i": best_inner_i,
                        "best_inner_val_loss": best_val,
                        "selected_seeds": [model_info["seed"] for model_info in best_seed_models],
                        "selected_inner_val_losses": [model_info["val_loss"] for model_info in best_seed_models],
                        "retrain_final": retrain_final_now,
                        "top_k_ensemble": top_k_ensemble,
                        "effective_top_k": 1 if selection_mode == "single" else top_k_ensemble,
                        "model_paths": [info["model_path"] for info in final_model_infos],
                        "metadata_path": str(metadata_path),
                    }
                )

                result_path = result_dir_now / f"results_{participant_id}_{outcome}.pth"
                result_path.parent.mkdir(parents=True, exist_ok=True)
                torch.save(results_by_variant[variant_tag], result_path)

        return results_by_variant

    print("LoRA-only path enabled for Chronos-2.")
    print("LoRA settings:", LORA_SETTINGS)
    print("AUG_TECHNIQUES overridden to:", AUG_TECHNIQUES)
```

```python
import copy
import math
from pathlib import Path
from itertools import product

import numpy as np
import pandas as pd
import torch
import torch.optim as optim

from torch.utils.data import DataLoader
from sklearn.metrics import mean_squared_error, r2_score

# =============================================================================
# Global settings
# =============================================================================

BASE_DIR = Path(
    "/content/drive/MyDrive/Colab_Notebooks/Anna_JacobsonColabDrive/"
    "SensingIndividualized"
)

BASE_DATA_DIR = BASE_DIR / "deeplearning_data" / "pred"
EMBED_DIR = BASE_DIR / "embeddings_pred"

MODEL_TYPE =  "Model2_LoRA_Combined"



AUG_TECHNIQUES = ["none"]


# Always save all four final versions
FINAL_VARIANTS = [
    {
        "variant_tag": "no_retrain_single",
        "selection_mode": "single",
        "retrain_final": False,
    },
    # {
    #     "variant_tag": "retrain_single",
    #     "selection_mode": "single",
    #     "retrain_final": True,
    # },
    # {
    #     "variant_tag": "no_retrain_ensemble",
    #     "selection_mode": "ensemble",
    #     "retrain_final": False,
    # },
    # {
    #     "variant_tag": "retrain_ensemble",
    #     "selection_mode": "ensemble",
    #     "retrain_final": True,
    # },
]

if not globals().get("USE_CHRONOS_LORA", False):
    raise RuntimeError(
        "This notebook is configured for LoRA-only runs. Execute the LoRA override cell before the main loop."
    )

print("Notebook mode: LoRA-only")


# =============================================================================
# Hyperparameter setup
# =============================================================================


participant_ids = ["00368","00179"]

outcomes = ["negative", "cognitive", "positive"]

outer_cfg = dict(
    train_size=135,
    test_size=3,
    step_size=3
)

inner_cfg = dict(
    val_size=21,
    step_size=114
)

fusion_opts = [
   # {"bottleneck": 8, "dropout": 0},
    {"bottleneck": 16, "dropout": 0.01},
    {"bottleneck": 32, "dropout": 0.01},
   # {"bottleneck": 64, "dropout": 0.01},
]

lora_lr_opts = [1e-5]
head_lr_opts = [0.0004]
alpha_opts = [0.02]

num_seed_runs = 1

# Used only for ensemble variants.
top_k_ensemble = 3

retrain_epoch_scale = 0.7
retrain_lr = 1e-4

hyperparams_grid = []

for f, lora_lr, head_lr, alpha in product(
    fusion_opts,
    lora_lr_opts,
    head_lr_opts,
    alpha_opts,
):
    hyperparams_grid.append(
        {
            "fusion": f,
            "lora_lr": lora_lr,
            "head_lr": head_lr,
            "alpha": alpha,
        }
    )

print(f"Total hyperparameter sets: {len(hyperparams_grid)}")
print(f"Seed runs per hyperparameter set: {num_seed_runs}")
print(f"Top ensemble members retained: {top_k_ensemble}")
print("Saving all variants:")
for item in FINAL_VARIANTS:
    print("  ", item["variant_tag"])

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

all_results = {}
all_test_losses = {}


# =============================================================================
# Participant/outcome loop
# =============================================================================

for participant_id in participant_ids:

    file_path = BASE_DATA_DIR / f"participant_{participant_id}_4h.npz"

    print("\n" + "=" * 80)
    print(f"Participant: {participant_id}")
    print("=" * 80)

    try:
        loaded = np.load(file_path, allow_pickle=True)
    except FileNotFoundError:
        print(f"File not found, skipping participant {participant_id}: {file_path}")
        continue

    datetimes = pd.to_datetime(loaded["datetimes"])
    order = np.argsort(datetimes.values)

    labels_all = loaded["labels"].astype(np.float32)

    # Combined head only
    features_combined = loaded["combined"]

    # Sort by datetime
    features_combined = features_combined[order]
    labels_all = labels_all[order]
    datetimes = datetimes[order]

    # Scale / round labels
    labels_all = (np.round(labels_all)).astype(np.float32) / 100

    # Remove first questionnaire
    features_combined = features_combined[1:]
    labels_all = labels_all[1:]
    datetimes = datetimes[1:]

    for outcome in outcomes:

        print("\n" + "-" * 80)
        print(f"Running participant {participant_id}, outcome: {outcome}")
        print("-" * 80)

        labels = get_labels_for_outcome(labels_all, outcome)

        for aug_technique in AUG_TECHNIQUES:

            print("\n" + "=" * 60)
            print(f"Augmentation: {aug_technique}")
            print("=" * 60)

            # Separate folders for each augmentation.
            folder_aug = "" if aug_technique == "none" else f"_{aug_technique}"

            MODEL_DIRS_AUG = {
                item["variant_tag"]: (
                    BASE_DIR
                    / f"final_models_{MODEL_TYPE}{folder_aug}_{item['variant_tag']}"
                )
                for item in FINAL_VARIANTS
            }

            RESULT_DIRS_AUG = {
                item["variant_tag"]: (
                    BASE_DIR
                    / "results_true_pred"
                    / f"{MODEL_TYPE}{folder_aug}_{item['variant_tag']}"
                )
                for item in FINAL_VARIANTS
            }

            for d in list(MODEL_DIRS_AUG.values()) + list(RESULT_DIRS_AUG.values()):
                d.mkdir(parents=True, exist_ok=True)

            try:
                results_by_variant = run_nested_cv_for_person(
                    features_combined=features_combined,
                    labels=labels,
                    participant_id=participant_id,
                    outcome=outcome,
                    outer_cfg=outer_cfg,
                    inner_cfg=inner_cfg,
                    device=device,
                    hyperparams_grid=hyperparams_grid,
                    model_dirs=MODEL_DIRS_AUG,
                    result_dirs=RESULT_DIRS_AUG,
                    embed_dir=EMBED_DIR,
                    final_variants=FINAL_VARIANTS,
                    max_epochs=100,
                    patience=20,
                    num_seed_runs=num_seed_runs,
                    top_k_ensemble=top_k_ensemble,
                    retrain_epoch_scale=retrain_epoch_scale,
                    retrain_lr=retrain_lr,
                    seed_base=42,
                    aug_technique=aug_technique,
                )

                for variant_tag, results in results_by_variant.items():

                    test_losses = [
                        mean_squared_error(item["y_test"], item["y_pred"])
                        for item in results
                    ]

                    mean_test_loss = np.mean(test_losses)

                    result_key = (
                        participant_id,
                        outcome,
                        aug_technique,
                        variant_tag,
                    )

                    all_results[result_key] = results
                    all_test_losses[result_key] = {
                        "per_outer_split": test_losses,
                        "mean": mean_test_loss,
                    }

                    print(
                        f"\n{aug_technique} / {variant_tag} "
                        "per-outer-split test losses:"
                    )
                    print(test_losses)

                    print(
                        f"{aug_technique} / {variant_tag} "
                        f"mean test loss: {mean_test_loss}"
                    )

            except Exception as e:

                print(
                    f"Error for participant {participant_id}, "
                    f"outcome {outcome}, augmentation {aug_technique}: {e}"
                )

                for item in FINAL_VARIANTS:
                    variant_tag = item["variant_tag"]

                    result_key = (
                        participant_id,
                        outcome,
                        aug_technique,
                        variant_tag,
                    )

                    all_results[result_key] = None
                    all_test_losses[result_key] = None
```
